```python
## Calculate RFM features for customers
## RFM means:
## Recency: How many days since the customer's last order.
## Frequency: How many distinct orders the customer placed.
## Monetary: Total purchase value of the customer.
```

```python
# Convert order date to datetime
orders["order_date"] = pd.to_datetime(orders["order_date"])

# Join orders with order items
sales = pd.merge(
    orders[["order_id", "customer_id", "order_date"]],
    order_items[["order_id", "total_sales"]],
    on="order_id",how="inner")

display(sales.head())
```

```python
# Fetch orders and items data
query = """
SELECT 
    o.customer_id,
    o.order_id,
    o.order_date,
    (oi.quantity * oi.list_price * (1 - oi.discount)) AS line_total
FROM orders o
JOIN order_items oi ON o.order_id = oi.order_id;
"""

df = pd.read_sql(query, engine)
df["order_date"] = pd.to_datetime(df["order_date"])
df.head()
```

```python
# Reference snapshot date (1 day after the latest order)
snapshot_date = df["order_date"].max() + pd.Timedelta(days=1)

# Group by customer
rfm = (
    df.groupby("customer_id")
    .agg(
        Recency=("order_date", lambda x: (snapshot_date - x.max()).days),
        Frequency=("order_id", "nunique"),
        Monetary=("line_total", "sum"),
    )
    .reset_index()
)

rfm.head()
```

```python
# Simple scoring using quartiles (1 to 4)
rfm["R_Score"] = pd.qcut(rfm["Recency"], q=4, labels=[4, 3, 2, 1]).astype(int)
rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"), q=4, labels=[1, 2, 3, 4]
).astype(int)
rfm["M_Score"] = pd.qcut(rfm["Monetary"], q=4, labels=[1, 2, 3, 4]).astype(int)

# Overall average score
rfm["Total_Score"] = (rfm["R_Score"] + rfm["F_Score"] + rfm["M_Score"]) / 3


# Map scores to simple segment labels
def get_segment(score):
    if score >= 3.5:
        return "Champions"
    elif score >= 2.5:
        return "Loyal Customers"
    elif score >= 1.75:
        return "At Risk"
    else:
        return "Lost"


rfm["segment_name"] = rfm["Total_Score"].apply(get_segment)
rfm[["customer_id", "Recency", "Frequency", "Monetary", "segment_name"]].head()
```

```python
# Select required columns matching your SQL table[cite: 1]
customer_segments = rfm[["customer_id", "segment_name"]]

# Save to SQL table (matches table created in SQL script)[cite: 1]
customer_segments.to_sql(
    name="customer_segments", con=engine, if_exists="append", index=False
)

print(
    f"Successfully exported {len(customer_segments)} records to customer_segments!"
)
```
